```python
import galsim 
```

```python
def draw_spergel(
    *,
    nu: float,
    f: float,
    hlr: float,
    e1: float,
    e2: float,
    psf_fwhm: float = 0.8,
):

    gal = galsim.Spergel(nu=nu, flux=f, half_light_radius=hlr)
    gal = gal.shear(g1=e1, g2=e2)

    psf = galsim.Gaussian(flux=1.0, fwhm=psf_fwhm)
    gal_conv = galsim.Convolve([gal, psf])
    return gal_conv
```

```python
for nu in [-0.6, -0.4, -0.2, 0.0, 0.1, 0.3, 0.5, 1.0, 2.0, 3.0]:
    gal = draw_spergel(nu=nu, f=1000, hlr=0.4, e1=0.1, e2=0.2)
    image = gal.drawImage(nx=63, ny=63, scale=0.2, offset=(0, 0))
    print(gal.getGoodImageSize(0.2))
print()
print(galsim.Image.good_fft_size(63))
```
